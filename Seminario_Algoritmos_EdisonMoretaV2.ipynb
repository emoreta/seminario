```python
# Calcular el costo de un conjunto de tomas
def calcular_costo(tomas_seleccionadas):
    actores_dia = [0] * 10
    for toma in tomas_seleccionadas:
        for i in range(len(actores_dia)):
            actores_dia[i] = max(actores_dia[i], toma[i])
    return sum(actores_dia)

# Generar todas las combinaciones posibles de 6 tomas
combinaciones = list(itertools.combinations(tomas, 6))

# Encontrar la combinación con el menor costo
costo_minimo = float('inf')
mejor_comb = None

for combinacion in combinaciones:
    costo = calcular_costo(combinacion)
    if costo < costo_minimo:
        costo_minimo = costo
        mejor_comb = combinacion

print("Costo mínimo:", costo_minimo)
print("Mejor combinación:", mejor_comb)
```

### Calcula la complejidad del algoritmo por fuerza bruta

La complejidad de este algoritmo de fuerza bruta es
𝑂\left(\binom{n}{k} \cdot n\right), donde 𝑛 es el número de tomas (30) y 𝑘 es el número de tomas por día (6). La función de combinación es factorial y se calcula como:

\[
\binom{n}{k} = \frac{n!}{k! \cdot (n - k)!}

Para 𝑛 = 30 y 𝑘 = 6, esto es aproximadamente 𝑂\left(\binom{30}{6}\right), que es una complejidad computacional muy alta y no es eficiente para valores grandes de 𝑛.

```python
# Número de actores
num_actores = 10

# Función para calcular el costo de un conjunto de tomas
def calcular_costo(tomas_seleccionadas):
    actores_dia = [0] * num_actores
    for toma in tomas_seleccionadas:
        for i in range(len(actores_dia)):
            actores_dia[i] = max(actores_dia[i], toma[i])
    return sum(actores_dia)

```

```python
def calcular_costo_minimo(tomas):
    dp = {}
    dp[0] = 0

    for i in range(len(tomas)):
        for estado in list(dp.keys()):
            nuevo_estado = estado | (1 << i)
            if bin(nuevo_estado).count('1') <= 6:
                tomas_seleccionadas = [tomas[j] for j in range(len(tomas)) if (nuevo_estado & (1 << j))]
                costo = calcular_costo(tomas_seleccionadas)
                if nuevo_estado not in dp or dp[nuevo_estado] > costo:
                    dp[nuevo_estado] = costo

    return min([dp[estado] for estado in dp if bin(estado).count('1') == 6])

print("Costo mínimo (Programación dinámica):", calcular_costo_minimo(tomas))
```

El enfoque de programación dinámica es más eficiente porque evita recalcular costos para las mismas combinaciones de tomas múltiples veces. Al almacenar los costos en una tabla y construir la solución óptima de manera incremental, reducimos significativamente la redundancia en los cálculos.

# Análisis de la complejidad

## El algoritmo de programación dinámica consta de dos partes principales:

La inicialización del array dp de tamaño 2^n, donde n es el número de tomas.
El cálculo del costo mínimo para cada estado del array dp.
Complejidad de la inicialización

La complejidad de la inicialización del array dp es O(2^n), ya que se necesita crear un array de tamaño 2^n y asignar un valor inicial a cada elemento.

## Complejidad del cálculo del costo mínimo

El cálculo del costo mínimo se realiza mediante un bucle que itera sobre todos los estados del array dp. Para cada estado, se calcula el costo mínimo de la siguiente manera:

Se seleccionan las tomas que corresponden al estado actual.
Se calcula el costo de la combinación de tomas seleccionadas.
Se actualiza el valor del array dp para el estado actual con el costo mínimo.
La complejidad de este bucle es O(2^n), ya que se necesita iterar sobre todos los estados del array dp.

## Complejidad del cálculo del costo

El cálculo del costo se realiza mediante la función calcular_costo, que tiene una complejidad de O(m), donde m es el número de actores.

## Complejidad total

La complejidad total del algoritmo de programación dinámica es la suma de las complejidades de la inicialización y del cálculo del costo mínimo:

O(2^n) + O(2^n) * O(m) = O(2^n) + O(2^n * m)

Simplificando la expresión, obtenemos:

O(2^n * m)

## Análisis de la complejidad en función de los parámetros

La complejidad del algoritmo de programación dinámica depende de dos parámetros: n (el número de tomas) y m (el número de actores).

Cuando n es pequeño, la complejidad del algoritmo es baja, ya que se necesita iterar sobre un número pequeño de estados.
Cuando n es grande, la complejidad del algoritmo es alta, ya que se necesita iterar sobre un número grande de estados.
Cuando m es pequeño, la complejidad del algoritmo es baja, ya que se necesita realizar un número pequeño de cálculos para calcular el costo.
Cuando m es grande, la complejidad del algoritmo es alta, ya que se necesita realizar un número grande de cálculos para calcular el costo.
Conclusión

La complejidad del algoritmo de programación dinámica es O(2^n * m), donde n es el número de tomas y m es el número de actores. La complejidad del algoritmo depende de los valores de n y m, y puede variar desde baja a alta dependiendo de los valores de estos parámetros.

Es importante tener en cuenta que la complejidad del algoritmo de programación dinámica puede ser alta para valores grandes de n y m, por lo que es importante utilizar técnicas de optimización adicionales para reducir la complejidad del algoritmo en estos casos.

```python
# Generar todos los partidos posibles entre los equipos
def generar_partidos(equipos_A, equipos_B, equipos_C):
    partidos = []
    for equipo_A in range(equipos_A):
        for equipo_B in range(equipos_B):
            partidos.append(('A', 'B'))
    for equipo_A in range(equipos_A):
        for equipo_C in range(equipos_C):
            partidos.append(('A', 'C'))
    for equipo_B in range(equipos_B):
        for equipo_C in range(equipos_C):
            partidos.append(('B', 'C'))
    return partidos

# Calcular la audiencia total para una asignación de horarios
def calcular_audiencia_total(asignacion_horarios, partidos):
    audiencia_total = 0
    for partido, horario in asignacion_horarios.items():
        categoria_1, categoria_2 = partido
        audiencia = audiencia_base.get((categoria_1, categoria_2), 0)
        audiencia_total += audiencia * horarios[horario] * (1 - reduccion_coincidencias.get(partidos.index(partido), 0))
    return audiencia_total
```

```python
def procesar_bloque(bloque, partidos, horarios):
    max_local = 0
    mejor_local = None
    for combinacion in bloque:
        asignacion_horarios = dict(zip(partidos, combinacion))
        audiencia_total = calcular_audiencia_total(asignacion_horarios, partidos)
        if audiencia_total > max_local:
            max_local = audiencia_total
            mejor_local = asignacion_horarios
    return max_local, mejor_local
# Resolver por fuerza bruta usando bloques iterativos
def resolver_por_fuerza_bruta(equipos_categoria_A, equipos_categoria_B, equipos_categoria_C, bloque_tamano=1000):
    partidos = generar_partidos(equipos_categoria_A, equipos_categoria_B, equipos_categoria_C)
    max_audiencia_total = 0
    mejor_asignacion_horarios = None

    # Generar todas las combinaciones posibles de horarios en bloques
    combinaciones = itertools.product(horarios.keys(), repeat=len(partidos))

    with concurrent.futures.ProcessPoolExecutor() as executor:
        while True:
            bloque = list(itertools.islice(combinaciones, bloque_tamano))
            if not bloque:  # Si no quedan combinaciones, detener
                break

            # Enviar bloque al procesador
            futures = [executor.submit(procesar_bloque, bloque, partidos, horarios)]
            for future in concurrent.futures.as_completed(futures):
                max_local, mejor_local = future.result()
                if max_local > max_audiencia_total:
                    max_audiencia_total = max_local
                    mejor_asignacion_horarios = mejor_local

    return mejor_asignacion_horarios, max_audiencia_total
```

```python
def algoritmo_voraz(partidos, horarios, audiencia_base, reduccion_coincidencias):
    asignacion = []
    coincidencias = {h: 0 for h in horarios.keys()}
    partidos_ordenados = sorted(partidos, key=lambda x: audiencia_base[tuple(sorted(x))], reverse=True)

    for partido in partidos_ordenados:
        mejor_horario = None
        mejor_audiencia = 0

        for horario in horarios.keys():
            base_audiencia = audiencia_base[tuple(sorted(partido))]
            coef_horario = horarios[horario]
            audiencia = base_audiencia * coef_horario

            if coincidencias[horario] > 0:
                reduccion = reduccion_coincidencias.get(coincidencias[horario], 0.8)  # Utiliza un valor por defecto si el número de coincidencias es mayor que el máximo
                audiencia -= audiencia * reduccion

            if audiencia > mejor_audiencia:
                mejor_audiencia = audiencia
                mejor_horario = horario

        asignacion.append((partido, mejor_horario))
        coincidencias[mejor_horario] += 1

    audiencia_total = sum([audiencia_base[tuple(sorted(partido))] * horarios[horario] for partido, horario in asignacion])

    for h, num_coincidencias in coincidencias.items():
        if num_coincidencias > 1:
            reduccion = reduccion_coincidencias.get(num_coincidencias - 1, 0.8)  # Utiliza un valor por defecto si el número de coincidencias es mayor que el máximo
            audiencia_total -= audiencia_total * reduccion

    return asignacion, audiencia_total

mejor_asignacion_voraz, mejor_audiencia_voraz = algoritmo_voraz(partidos, horarios, audiencia_base, reduccion_coincidencias)

print("Mejor audiencia total (voraz):", mejor_audiencia_voraz)
print("Mejor asignación de horarios (voraz):", mejor_asignacion_voraz)
```

El problema consiste en analizar el siguiente problema y diseñar un algoritmo que lo resuelva
Disponemos de las 9 cifras del 1 al 9 (excluimos el cero) y de los 4 signos básicos de las
operaciones fundamentales: suma(+), resta(-), multiplicación(*) y división(/)
Debemos combinarlos alternativamente sin repetir ninguno de ellos para obtener una
cantidad dada. Un ejemplo sería para obtener el 4:
4+2-6/3*1 = 4
Debe analizarse el problema para encontrar todos los valores enteros posibles planteando las
siguientes cuestiones:
- ¿Qué valor máximo y mínimo se pueden obtener según las condiciones del problema?
- ¿Es posible encontrar todos los valores enteros posibles entre dicho mínimo y máximo ?

```python
import itertools

# Cifras y operaciones disponibles
cifras = list('123456789')
operaciones = list('+-*/')

# Generar todas las permutaciones posibles de cifras y operaciones
permutaciones_cifras = list(itertools.permutations(cifras))
permutaciones_operaciones = list(itertools.permutations(operaciones))

# Generar todas las expresiones posibles
def generar_expresiones():
    expresiones = []
    for perm_cifras in permutaciones_cifras:
        for perm_operaciones in permutaciones_operaciones:
            expr = []
            for i in range(len(perm_operaciones)):
                expr.append(perm_cifras[i])
                expr.append(perm_operaciones[i])
            expr.append(perm_cifras[-1])
            expresiones.append(''.join(expr))
    return expresiones

expresiones = generar_expresiones()
```

```python
def evaluar_expresiones(expresiones):
    resultados = {}
    for expr in expresiones:
        try:
            resultado = eval(expr)
            if resultado == int(resultado):
                resultados[expr] = int(resultado)
        except ZeroDivisionError:
            continue
    return resultados

resultados = evaluar_expresiones(expresiones)
```

```python
valores = list(resultados.values())
valor_minimo = min(valores)
valor_maximo = max(valores)
valores_posibles = set(range(valor_minimo, valor_maximo + 1))
valores_encontrados = set(valores)

todos_valores_posibles = valores_posibles.issubset(valores_encontrados)

print("Valor mínimo:", valor_minimo)
print("Valor máximo:", valor_maximo)
print("Todos los valores enteros posibles están presentes:", todos_valores_posibles)
```
